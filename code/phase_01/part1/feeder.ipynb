```python
import time
import os

SOURCE_FILE = "/home/reza/HW4/traffic.csv"
DEST_DIR = "/home/reza/HW4/stream_input"
CHUNK_SIZE = 5000
INTERVAL = 5

def simulate_streaming():
    if not os.path.exists(DEST_DIR):
        os.makedirs(DEST_DIR)

    with open(SOURCE_FILE, 'r') as f:
        header = f.readline()
        batch_lines = []
        batch_counter = 1
        
        for line in f:
            batch_lines.append(line)
            
            if len(batch_lines) >= CHUNK_SIZE:
                output_file = f"{DEST_DIR}/batch_{batch_counter}.csv"
                with open(output_file, 'w') as out_f:
                    out_f.writelines(batch_lines)
                
                print(f"Batch {batch_counter} moved.")
                batch_lines = []
                batch_counter += 1
                time.sleep(INTERVAL)

        if batch_lines:
            output_file = f"{DEST_DIR}/batch_{batch_counter}.csv"
            with open(output_file, 'w') as out_f:
                out_f.writelines(batch_lines)
            print(f"Final Batch {batch_counter} moved.")

if __name__ == "__main__":
    simulate_streaming()
```
