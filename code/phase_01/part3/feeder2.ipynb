```python
import time
import os
import shutil


SOURCE_FILE = "/home/reza/HW4/camera_passes.csv"
DEST_DIR = "/home/reza/HW4/stream_input"
TEMP_DIR = "/home/reza/HW4/stream_stage"

CHUNK_SIZE = 1000  
INTERVAL = 5

def simulate_streaming():
    if not os.path.exists(DEST_DIR):
        os.makedirs(DEST_DIR)
    if not os.path.exists(TEMP_DIR):
        os.makedirs(TEMP_DIR)

    print(f"Starting Feeder from {SOURCE_FILE}...")
    
    with open(SOURCE_FILE, 'r') as f:
        header = f.readline() 
        batch_lines = []
        batch_counter = 1
        
        for line in f:
            batch_lines.append(line)
            
            if len(batch_lines) >= CHUNK_SIZE:
                temp_file = f"{TEMP_DIR}/batch_{batch_counter}.tmp"
                final_file = f"{DEST_DIR}/batch_{batch_counter}.csv"
                
                with open(temp_file, 'w') as out_f:
                    out_f.writelines(batch_lines)
                
                shutil.move(temp_file, final_file)
                print(f"Batch {batch_counter} moved.")
                
                batch_lines = []
                batch_counter += 1
                time.sleep(INTERVAL)

        if batch_lines:
            temp_file = f"{TEMP_DIR}/batch_{batch_counter}.tmp"
            final_file = f"{DEST_DIR}/batch_{batch_counter}.csv"
            with open(temp_file, 'w') as out_f:
                out_f.writelines(batch_lines)
            shutil.move(temp_file, final_file)
            print(f"Final Batch {batch_counter} moved.")

if __name__ == "__main__":
    simulate_streaming()
```
